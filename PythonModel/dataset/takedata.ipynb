```python
import pandas as pd

# cargar csv
df = pd.read_csv("test_dataset.csv")

# separar clases
class1 = df[df["class"] == 1].iloc[:100, :-1]   # primeras 100 filas clase 1
class0 = df[df["class"] == 0].iloc[:100, :-1]   # primeras 100 filas clase 0

# convertir a listas planas
gamma = class1.to_numpy().flatten()
neutron = class0.to_numpy().flatten()

def write_c_array(name, data, width=161):
    print(f"int {name}[100 * {width}] = {{")
    
    for i in range(0, len(data), width):
        row = data[i:i+width]
        line = ",".join(str(int(v)) for v in row)
        print(f"\t\t{line},")
    
    print("};\n")

# imprimir arrays en formato C
write_c_array("tracesGamma", gamma)
write_c_array("tracesNeutron", neutron)
```
